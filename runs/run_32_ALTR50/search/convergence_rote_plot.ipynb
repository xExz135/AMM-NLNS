```python
import pandas as pd
import matplotlib.pyplot as plt

# Read convergence log
df = pd.read_csv("convergence_cpu_SpawnPoolWorker-1.csv", comment="#", names=["iteration","cost"])

# Create plot
plt.figure(figsize=(8,5))
plt.plot(df["iteration"], df["cost"])

plt.xlabel("Iteration")
plt.ylabel("Mean Cost")
plt.title("NLNS Convergence Curve (Chunk Size 32, ALTR_C_20)")
plt.grid(True)

plt.tight_layout()
plt.savefig("convergence_curve.png", dpi=300)
plt.show()
```

```python
initial = df["cost"].iloc[0]

df["improvement_pct"] = (
    (initial - df["cost"]) / initial
) * 100

plt.figure(figsize=(8,5))
plt.plot(df["iteration"], df["improvement_pct"])

plt.xlabel("Iteration")
plt.ylabel("Improvement (%)")
plt.title("Improvement Relative to Initial Solution")
plt.grid(True)

plt.tight_layout()
plt.savefig("improvement_curve.png", dpi=300)
plt.show()
```
